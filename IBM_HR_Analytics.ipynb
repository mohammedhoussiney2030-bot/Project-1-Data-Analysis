```python
IBM_HR_Analytics.drop(columns=["EmployeeCount","StandardHours","Over18"])
IBM_HR_Analytics.shape
```

```python
# Option 1 (recommended): assign the result back
df = IBM_HR_Analytics.copy()
df = df.drop(columns=["EmployeeCount", "StandardHours", "Over18","EmployeeNumber"])
df.shape
```

```python
df["AttritionFlag"] = (df["Attrition"] == "Yes").astype(int)
print(df.shape)
print(df["AttritionFlag"].mean())
```

```python
df["AgeGroup"] = pd.cut(df["Age"], bins=[17,25,35,45,55,100],labels=["18-25","26-35","36-45","46-55","56+"])
df["TenureGroup"] = pd.cut(df["YearsAtCompany"], bins=[-1,2,5,10,50], labels=["0-2 yrs","3-5 yrs","6-10 yrs","10+ yrs"])
# Readable satisfaction labels
satisfaction_map = {1: "Low", 2: "Medium", 3: "High", 4: "Very High"}
for col in ["JobSatisfaction", "EnvironmentSatisfaction", "WorkLifeBalance"]:
    df[col + "_Label"] = df[col].map(satisfaction_map)

print(df.shape)
```

```python
def attrition_rate(column):
    result = (df.groupby(column, observed=True)["AttritionFlag"]
                .agg(["mean", "sum", "count"]))
    result["mean"] = (result["mean"] * 100).round(1)
    result.columns = ["Attrition %", "Left", "Total"]
    return result.sort_values("Attrition %", ascending=False)

attrition_rate("OverTime")
```

```python
for col in ["Department", "JobRole", "AgeGroup", "MaritalStatus", "TenureGroup"]:
    print(f"\n--- Attrition by {col} ---")
    display(attrition_rate(col))
```

```python
import os

sns.set_theme(style="whitegrid")
company_avg = df["AttritionFlag"].mean() * 100

fig, axes = plt.subplots(1, 3, figsize=(16, 4.8))

for ax, col in zip(axes, ["OverTime", "AgeGroup", "TenureGroup"]):
    data = attrition_rate(col).reset_index()
    data[col] = data[col].astype(str)
    colors = ["#C0392B" if v > company_avg + 3 else "#2E75B6" for v in data["Attrition %"]]
    sns.barplot(data=data, x=col, y="Attrition %", ax=ax, palette=colors,
                hue=col, legend=False, dodge=False)
    for container in ax.containers:
        ax.bar_label(container, fmt="%.1f%%", padding=3)
    ax.axhline(company_avg, color="gray", linestyle="--", label=f"Company avg {company_avg:.1f}%")
    ax.set_title(f"Attrition by {col}", fontweight="bold")
    ax.set_xlabel("")
    ax.set_ylabel("Attrition %")
    ax.set_ylim(0, data["Attrition %"].max() * 1.15)
    ax.legend()

fig.suptitle("Who leaves? Overtime, young age, and short tenure stand out", fontsize=14, fontweight="bold")
fig.text(0.01, -0.02, f"Dashed line = company average ({company_avg:.1f}%). Red = more than 3 points above average.", fontsize=9)
plt.tight_layout()

os.makedirs("../images", exist_ok=True)
plt.savefig("../images/attrition_drivers.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
role_ot = (df.groupby(["JobRole", "OverTime"])["AttritionFlag"]
             .agg(rate="mean", n="count").reset_index())
role_ot["rate"] = role_ot["rate"] * 100

order = list(role_ot[role_ot["OverTime"] == "Yes"]
             .sort_values("rate", ascending=False)["JobRole"])

fig, ax = plt.subplots(figsize=(11, 6.5))
sns.barplot(data=role_ot, y="JobRole", x="rate", hue="OverTime", order=order,
            hue_order=["No", "Yes"], palette={"No": "#2E75B6", "Yes": "#C0392B"}, ax=ax)

for container, ot in zip(ax.containers, ["No", "Yes"]):
    labels = []
    for role in order:
        row = role_ot[(role_ot["JobRole"] == role) & (role_ot["OverTime"] == ot)].iloc[0]
        labels.append(f"{row['rate']:.1f}%" + ("*" if row["n"] < 30 else ""))
    ax.bar_label(container, labels=labels, padding=3)

ax.set_xlim(0, 80)
ax.set_xlabel("Attrition %")
ax.set_ylabel("")
ax.set_title("Overtime raises attrition in almost every role", fontsize=14, fontweight="bold")
fig.text(0.01, -0.02, "* Group has fewer than 30 employees; treat with caution.", fontsize=9)
plt.tight_layout()
plt.savefig("../images/attrition_role_overtime.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
order = ["Low", "Medium", "High", "Very High"]
company_avg = df["AttritionFlag"].mean() * 100

fig, axes = plt.subplots(1, 3, figsize=(16, 4.8), sharey=True)
for ax, col in zip(axes, ["JobSatisfaction_Label", "EnvironmentSatisfaction_Label", "WorkLifeBalance_Label"]):
    data = attrition_rate(col).reindex(order).reset_index()
    colors = ["#C0392B" if v > company_avg + 3 else "#2E75B6" for v in data["Attrition %"]]
    sns.barplot(data=data, x=col, y="Attrition %", palette=colors, hue=col, legend=False, ax=ax)
    for container in ax.containers:               # label every bar, not just the first
        ax.bar_label(container, fmt="%.1f%%", padding=3)
    ax.axhline(company_avg, color="gray", linestyle="--")
    ax.set_title(col.replace("_Label", ""), fontweight="bold")
    ax.set_xlabel("")
    ax.set_ylim(0, 37)

fig.suptitle("Leavers earn less overall, but mostly because they are concentrated in entry-level roles", fontsize=14, fontweight="bold")
fig.text(0.01, -0.02, f"Dashed line = company average ({company_avg:.1f}%). Red = more than 3 points above average.", fontsize=9)
plt.tight_layout()
plt.savefig("../images/satisfaction.png", dpi=150, bbox_inches="tight")
plt.show()
```

# Step 7:  "Employee Attrition Analysis: Key Findings and Recommendations".

## Key findings:

- **Overall:**   (237 of 1,470 left, 16.1%)
- **Overtime:**  (30.5% vs 10.4%; 8 of 9 roles)
- **Early career:**   (Level 1 = about 60% of leavers; 29.8% in first 2 years)
- **Satisfaction:**   (Low: work-life 31.2%, environment 25.4%, job 22.8%)
- **Pay:**   (matters at entry level: low-paid third 33.7% vs high-paid third 18.9%)
- **Highest-risk profile:**  (65 employees, 61.5% vs 14.0% for everyone else)

## Recommendations:

Review overtime workloads, starting with Sales Representatives and Lab Technicians.
Build a stronger onboarding and check-in program for the first 2 years.
Review entry-level pay bands.
Run a short survey for employees who rate environment or work-life balance as Low.

## Limitations:
the data shows association, not causation. Some groups are small. The profile was chosen after looking at the data.
